# Day 1 — Data Understanding

Load the uploaded CSVs, inspect shape/columns/types/missing values/duplicates, and identify possible keys.

```python
import pandas as pd
from pathlib import Path
RAW = Path("../data/raw")
files = list(RAW.glob("*.csv"))
for f in files:
    df = pd.read_csv(f)
    print(f.name, df.shape)
    print(df.columns.tolist())
    print(df.isna().sum().sort_values(ascending=False).head(10))
```

Important finding: employee datasets use different ID namespaces, so do not merge them without proof.
